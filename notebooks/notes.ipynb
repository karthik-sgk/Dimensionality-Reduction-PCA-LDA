<b><i>Dimensionality Reduction:</b></i>is the process of reducing the number of input features in a dataset while preserving as much useful information as possible.

For example, a dataset with 100 features may be reduced to 10 features while still retaining the important structure of the data.

<b>Why reduce dimensionality?</b>

1. Improves model performance
    * High-dimensional data may lead to overfitting.
    * Reducing dimensions helps models generalize better.
2. Reduces computational cost
    * Fewer features means faster training and inference.
3. Removes redundancy
    * Many features may contain duplicate or correlated information.
4. Visualization
    * Dimensionality reduction allows high-dimensional data to be visualized in 2D or 3D space.

<b>Types of Dimensionality Reduction:</b>   

1. Feature Selection
    * Selecting a subset of original features.

2. Feature Extraction
    * Creating new features as combinations of existing ones.
    * Examples:
        * PCA
        * LDA
        
Overall, dimensionality reduction is an important step in the data preprocessing pipeline that can help improve model performance, reduce computational cost, and facilitate data visualization. It is important to choose the appropriate technique based on the specific characteristics of the dataset and the goals of the analysis.
</div>

<b>Curse of Dimensionality:</b>
The curse of dimensionality refers to the problems that arise when working with data in high-dimensional spaces. As the number of features increases, the volume of the feature space grows exponentially, making data points sparse. This sparsity causes machine learning algorithms to struggle in identifying meaningful patterns.

Problems Caused by High Dimensions

<b>1. Data Sparsity</b>

In high dimensions, data points become far apart from each other.
This makes it harder for algorithms like:
* KNN
* Clustering
* Density estimation

to find meaningful neighbors.

<b>2. Distance Metrics Become Less Useful</b>

In high-dimensional spaces, the difference between the nearest and farthest neighbor distances becomes very small. This makes algorithms that rely on distance measures less effective.

<b>3. Increased Computational Cost</b>
More features lead to:
* Larger matrices
* Slower training
* Higher memory usage

<b>4. Overfitting</b>

With many features, models can start fitting noise instead of real patterns. This reduces generalization to unseen data.

<b>How Dimensionality Reduction Helps</b>

Dimensionality reduction techniques like PCA and LDA help by:
* Removing redundant features
* Preserving important information
* Reducing computational complexity
* Improving model generalization

<b><i>Linear Dependence and Independence</b></i>

Definition
* A set of vectors {v_1,v_2,…,v_n }is said to be:
* <b> Linearly dependent </b> if at least one vector can be written as a linear combination of the others.
	
	c_1 v_1+c_2 v_2+⋯+c_n v_n=0 (with some c_i≠0)  
* <b>Linearly independent</b> if the only solution to the above is the trivial one:
	
	c_1=c_2=⋯=c_n=0

<b>Intuition</b>
* <b>Independent set</b> = no redundancy; every vector adds unique information.
* <b>Dependent set</b> = at least one vector is redundant.

<b>Examples</b>

<b>Case 1: Two dependent features</b>
* x_2=3x_1

⇒ x_1,x_2are linearly dependent (scalar multiples).

<b>Case 2: General dependency</b>
* x_3=2x_1+5x_2

⇒ x_1,x_2,x_3are linearly dependent, because x_3adds no new information.

<b>Case 3: Independent features</b>
* In R^3, the set {(1,0,0),(0,1,0),(0,0,1)}is linearly independent. None can be written in terms of the others.

<b><i>Why It Matters in ML</b></i>

If features are linearly dependent:
* They provide duplicate information
* This increases dimensionality without adding value.

Dimensionality reduction techniques like PCA remove such redundancy.

<b>Variance and Covariance</b>
<br></br>
<b><i><u>Variance</b></i></u>

Variance measures how much a variable spreads around its mean.

$\text{Var}(X) = \frac{1}{n-1} \sum​_{i=1}^n (X_i​−μ)^2$

where:

$ X_i $: Individual data points.

$ μ $: Mean of the data.

Interpretation:
* High variance means data points are widely spread out.
* Low variance means data points are close to the mean.

<br></br>
<b><i><u>Covariance</b></i></u>

Covariance measures how two variables change together.
	
$\text{Cov}(X, Y) = \frac{1}{n-1} \sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y})$

where:

$ X_i, Y_i $: Individual data points.

$ \bar{X}, \bar{Y} $: Means of $ X $ and $ Y $.

Interpretation:
* Positive covariance: X and Y tend to increase together.
* Negative covariance: X increases while Y decreases.
* Zero covariance: X and Y are uncorrelated.
</div>


<div class="alert alert-success">

<b>Covariance Matrix</b>: The covariance matrix is a square matrix that contains the covariances between pairs of features in a dataset. Each element (i,j) of the covariance matrix represents the covariance between feature i and feature j. The diagonal elements represent the variance of each feature, while the off-diagonal elements represent the covariance between different features.

Example covariance matrix for three features:

$\text{Cov} = \begin{bmatrix}
\text{Var}(X_1) & \text{Cov}(X_1, X_2) & \text{Cov}(X_1, X_3) \\
\text{Cov}(X_2, X_1) & \text{Var}(X_2) & \text{Cov}(X_2, X_3) \\
\text{Cov}(X_3, X_1) & \text{Cov}(X_3, X_2) & \text{Var}(X_3)
\end{bmatrix}$

As the covariance matrix captures the relationships between all features in the dataset.

To identify the directions that maximize variance, we compute the eigenvectors and eigenvalues of this covariance matrix.
</div>

<b>Eigenvectors and Eigenvalues Intuition</b>

Eigenvectors and Eigenvalues are fundamental concepts in linear algebra that help us understand how transformations affect data.

They play a key role in Principal Component Analysis (PCA).

Intuition

When a matrix transforms a vector, the vector usually changes both direction and magnitude. However, some special vectors only scale in magnitude but do not change direction. These special vectors are called eigenvectors. The amount by which they scale is called the eigenvalue. Meaning: vecors can either stretch or shrink, but they do not rotate.

Mathematical Definition

For a square matrix A, if there exists a non-zero vector v and a scalar λ such that:

<b>A * v = λ * v</b>

Then v is an eigenvector of A, and λ is the corresponding eigenvalue.

This wquations means that when we apply the transformation A to the vector v, it only gets scaled by λ, without changing its direction.

<b>Geometric Interpretation</b>

Eigenvectors represent important directions in the data space.

Eigenvalues represent how much variance exists along those directions.

<b>Why Eigenvectors Matter for PCA</b>

PCA computes the eigenvectors of the covariance matrix.

These eigenvectors represent the principal components, which are the directions where the data varies the most.

The corresponding eigenvalues indicate how much variance each principal component captures.

<b>Example Interpretation</b>

If PCA produces:

Component	Eigenvalue	Meaning

PC1	           5.2	    Captures most variance

PC2	           1.1	    Captures less variance

We select the components with the largest eigenvalues.

<b>Principal Component Analysis (PCA)</b>

It is an unsupervised dimensionality reduction technique used in machine learning to identify relevant features and remove redundant or irrelevant ones from a dataset. The core objective of the PCA algorithm is to transform a given dataset with a certain number of features (n) into a new dataset with a smaller, user-defined number of features, known as principal components (p).

These <b>PCA</b> components:
* Are linear combinations of original features
* Are orthogonal (uncorrelated) to each other
* Capture the maximum variance

<b>Geometric Intuition</b>
* PCA finds a new coordinate system where:
* The first axis captures maximum variance
* The second axis captures the next largest variance
* And so on.

<b>PCA</b> works by computing the eigenvectors of the covariance matrix and selecting the directions with the largest eigenvalues, which correspond to the directions of maximum variance.

<b>PCA is one of the most widely used dimensionality reduction techniques </b> in machine learning and data science.

It is commonly used for:
- Data compression
- Noise reduction
- Feature extraction
- Data visualization

PCA Algorithm Steps:

1.  <b><i>Standardize the Data: </b></i>
Scale all variables so they are on the same range by subtracting the mean and dividing by the standard deviation. This prevents features with larger scales from dominating the analysis.

2. <b><i>Compute the Covariance Matrix: </b></i>
Calculate the covariance matrix to understand how features relate to each other. If two variables are highly correlated, they carry overlapping information. The covariance matrix helps reveal these redundancies.

    The covariance between any two variables, xi and xj, is calculated as follows: 

    $\text{Cov}(x_i, x_j) = \frac{1}{N - 1} \sum_{1}^{N} (x_{i} - \bar{x}_i)(x_{j} - \bar{x}_j)$

    where
    * N is the total number of examples.
    * $\bar{x}_i$ is the mean of the i-th feature.
    * $\bar{x}_j$ is the mean of the j-th feature.
    * $x_i,x_j$ is the observation in dataset.

3. <b><i>Find Eigenvalues and Eigenvectors: </b></i>
From the covariance matrix, compute eigenvalues and eigenvectors.
This step is the core of the PCA algorithm, where the principal components are mathematically identified by decomposing the covariance matrix S. This decomposition yields two critical sets of values:  
    *	<b><i>Eigenvectors</b></i> give the directions of maximum variance (principal components).
    *	<b><i>Eigenvalues</b></i> indicate how much variance is captured by each component.

    A high <b>eigenvalue</b> means its <b>eigenvector</b> points in a direction of significant data spread.

4.	<b><i>Select Principal Components</b></i>
Sort the calculated eigenvalues in decreasing order. Select the top 'k' eigenvalues, where 'k' is the desired number of dimensions for the new feature subspace that capture the majority of the variance for dimensionality reduction.

5. <b><i>Determine Eigenvectors (W)</b></i>
Calculate the eigenvectors corresponding to the selected top 'k' eigenvalues. This is done by solving the equation 
$(SW^-1 * SB - λI) * W = 0$ for each selected λ. The resulting eigenvectors (W) form the basis of the new feature space.

6. <b><i>Form the New Feature Space</b></i> Project the original data onto the new, lower-dimensional space. This is achieved by taking the dot product of the original data matrix and the matrix of selected eigenvectors. The result is the new set of features.

</div>

<b>Why is the Covariance Matrix Important in PCA?</b>

In Principal Component Analysis (PCA), the covariance matrix is critical because it captures the relationships (correlations) and variances among features, which PCA uses to identify the directions (principal components) that maximize data variance. Here’s why it matters:

<b>Captures Feature Relationships:</b>

* The covariance matrix shows how features are correlated. High covariance between two features (e.g., height and weight) indicates redundancy, as they vary together.

* PCA uses these relationships to combine correlated features into fewer, uncorrelated principal components, reducing dimensionality while preserving information.


<b>Determines Variance Directions:</b>

* PCA seeks the directions (axes) in the data that capture the most variance, as these represent the most significant patterns.
The covariance matrix’s eigenvectors define these directions (principal components), and the eigenvalues indicate how much variance each direction captures.

* By performing eigendecomposition on the covariance matrix, PCA identifies the axes along which the data varies most.

<b>Enables Dimensionality Reduction:</b>

* The covariance matrix allows PCA to rank principal components by their eigenvalues (variance explained). You can select the top components (e.g., reducing from 3 features to 1) to retain most of the data’s variance, simplifying the dataset.
* For example, in the custom dataset from the previous response, the correlated features (Feature1, Feature2, Feature3) had a covariance matrix showing strong relationships, enabling PCA to project the data onto a single component capturing ~85% of the variance.

<b>Why Calculate the Covariance Matrix?</b>

* Mathematical Foundation: The covariance matrix is the basis for PCA’s eigendecomposition. The eigenvectors are the principal components (new axes), and the eigenvalues quantify how much variance each component explains.

* Redundancy Reduction: High off-diagonal values (covariances) indicate correlated features, which PCA combines into fewer components, eliminating redundancy.

* Optimization: By maximizing variance (via eigenvalues), PCA ensures the reduced dataset retains the most important information, making analysis faster and more efficient.

<b>Explained Variance in PCA</b>

<b><i>Explained Variance</b></i>

In Principal Component Analysis (PCA), each principal component captures a certain amount of the total variance present in the dataset.

The explained variance measures how much information (variance) each principal component retains from the original dataset.

Since <b>PCA</b> orders components by descending variance, the first principal component captures the <b>maximum variance</b>, the second captures the next highest variance, and so on.

<b>Relationship with Eigenvalues</b>

In PCA, the <b>eigenvalues</b> of the covariance matrix represent the variance captured by each principal component.
* Larger eigenvalue → component captures more variance
* Smaller eigenvalue → component captures less variance

If the covariance matrix produces eigenvalues:

| Principal Component | Eigenvalue	|
|---------------------|-------------|
| PC1                 | 	5.5		|
| PC2                 | 	2.4 	|
| PC3                 | 	0.5		|

Then <b>PC1</b> explains the largest portion of the dataset variance.

<b>Explained Variance Ratio</b>

The <b>explained variance ratio</b> represents the proportion of total variance captured by each component.

It is calculated as:

$\text{Explained Variance Ratio}_i = \frac{\lambda_i}{\sum_{j=1}^{n}\lambda_j}$

where:
* $\text{λ}_i$ = eigenvalue of the i-th principal component
* n = total number of components

<b><i>Example</b></i>

Suppose PCA produces the following eigenvalues:

| Principal Component | Eigenvalue	|
|---------------------|-------------|
| PC1                 | 	4		|
| PC2                 | 	2 		|
| PC3                 | 	1		|

<b>Total variance:</b> 4 + 2 + 1 = 7


<b>Explained variance ratio:</b>

| Principal Component | Explained Variance	|
|---------------------|--------------------	|
| PC1                 | 	4/7 ≈ 57%		|
| PC2                 | 	2/7 ≈ 29%		|
| PC3                 | 	1/7 ≈ 14%		|

This means the first two components capture: <b>57% + 29% = 86%</b> of the total dataset variance.

<b>Choosing the Number of Components</b>

One of the key uses of explained variance is to determine how many principal components to retain. A common strategy is to keep enough components to capture 90–95% of the total variance.

For example:
| Component | Cumulative Variance	|
|-----------|--------------------	|
| PC1       | 	57%		            |
| PC2       | 	86%		            |
| PC3       | 	100%		        |


If the goal is to <b>retain at least 90% variance</b>, then we would <b>select PC1 and PC2.</b>

For Visualization:
A Scree Plot is often used to visualize explained variance.

It plots:
* X-axis → principal components
* Y-axis → explained variance
The plot helps identify the elbow point, where adding more components provides diminishing returns.

A Scree Plot can be used to visualize the explained variance of each principal component and determine the optimal number of components.

<b>Why Explained Variance Matters</b>

Explained variance helps:

* Determine the optimal number of principal components
* Remove low-information dimensions
* Balance information retention vs dimensionality reduction

While PCA focuses on preserving the maximum variance in the dataset, some applications require maximizing the separation between classes. In such cases, Linear Discriminant Analysis (LDA) is often used.

<div class="alert alert-success">
<b>Linear Discriminant Analysis (LDA):</b> It is a supervised dimensionality reduction technique that aims to find a linear combination of features that best separates two or more classes in the data. It is commonly used for classification tasks and can be particularly effective when the classes are well-separated in the feature space.
</div>

<div class='alert alert-success'>
The core principle of LDA is based on two simultaneous optimization criteria:

1. <b><i>Maximizing the distance between the means of the classes</b></i>. This pushes the class clusters as far apart as possible in the lower-dimensional space.

2. <b><i>Minimizing the variation (scatter) within each class</b></i>. This makes each class cluster as compact as possible.The process involves a series of computational steps, beginning with calculating the mean for each class. Subsequently, within-class and between-class scatter matrices are computed. These matrices are then used to solve an eigenvalue problem, yielding eigenvectors that form the new axes (linear discriminants). By sorting the eigenvalues and selecting the top 'k' corresponding eigenvectors, the original data can be projected onto a new 'k'-dimensional subspace that preserves maximum class-discriminatory information.
</div>

<div class='alert alert-success'>
<b>LDA: Algorithm Steps:</b>

1. <b><i> Standardize the Data </b></i>
Scale all variables so they are on the same range by subtracting the mean and dividing by the standard deviation. This prevents features with larger scales from dominating the analysis.  

2. <b><i> Compute Class Means (μ)</b></i>
Calculate the mean vector for each class. For a class W1 with N1 samples, the mean μ1 is the sum of all data points in that class divided by N1. This is repeated for all classes.

3. <b><i>Calculate Covariance Matrices (S)</b></i>
For each class, compute its covariance matrix. The formula for the covariance matrix of class one (S1) is S1 = Σ (x - μ1)(x - μ1)T for all data points x belonging to that class.

4. <b><i>Compute Within-Class Scatter Matrix (SW)</b></i>
Calculate the sum of the covariance matrices for all classes. This matrix represents the total scatter, or variation, within all classes. For two classes, SW = S1 + S2.

5. <b><i>Compute Between-Class Scatter Matrix (SB)</b></i>
Calculate the scatter between the different classes.  
$S_B = \sum_{i=1}^{c} n_i (\mu_i - \mu) (\mu_i - \mu)^T$
where:

    * $ c $ is the number of classes,
    * $ n_i $ is the number of samples in class $ i $,
    * $ \mu_i $ is the mean vector of class $ i $,
    * $ \mu $ is the overall mean vector of the dataset,
    * $ (\mu_i - \mu) (\mu_i - \mu)^T $ is the outer product of the difference between the class mean and the overall mean.

6. <b><i>Calculate Eigenvalues (λ)</b></i>
Solve the generalized eigenvalue problem using the within-class and between-class scatter matrices. The equation is 
$(SW^-1 * SB) * W = λ * W$. The number of non-zero eigenvalues will be at most the number of classes minus one.

7. <b><i>Select Linear Discrminants</b></i>
Sort the calculated eigenvalues in decreasing order. Select the top 'k' eigenvalues, where 'k' is the desired number of dimensions for the new feature subspace (i.e., the number of linear discriminants).

8. <b><i>Determine Eigenvectors (W)</b></i>
Calculate the eigenvectors corresponding to the selected top 'k' eigenvalues. This is done by solving the equation 
$(SW^-1 * SB - λI) * W = 0$ for each selected λ. The resulting eigenvectors (W) form the basis of the new feature space.

9. <b><i>Form the New Feature Space</b></i> Project the original data onto the new, lower-dimensional space. This is achieved by taking the dot product of the original data matrix and the matrix of selected eigenvectors. The result is the new set of features, or linear discriminants.
